```python
# Load libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Create Sample Data
np.random.seed(613) # Set seed for reproducibility
x = np.arange(50) # Creates an array of 50 numbers from 0 to 49. This will be our x-axis values.
y = np.random.randint(0, 100, 50) # Creates an array of 50 random integers between 0 and 100. This will be our y-axis values.

# Create basic Scatter Plot
fig, ax = plt.subplots(figsize=(5,3)) # Create a figure and axes object. The figsize argument sets plot size(width = 5 inches, height = 3 inches).
ax.scatter(x, y) # Plots the scatter plot using x and y. Each (x[i], y[i]) pair will be plotted as a point on the graph.
```

```python
# Using the same data, create a line plot
fig, ax = plt.subplots(figsize=(5,3)) # Create a new figure and axes object for the line plot.
ax.plot(x, y) # Plots a line plot using x and y. The points will be connected by lines in the order they appear in the arrays.
```

```python
# Using the same data, creat a Histogram
# A histogram is a graphical representation of the distribution of a dataset. It groups the data into bins and counts the number of data points that fall into each bin.

fig, ax = plt.subplots(figsize=(5,3)) # Create a new figure and axes
ax.hist(y, bins=10) # Plots a histogram of the y values. The bins argument specifies the number of bins to use for the histogram.
```

```python
# Create a bar plot Histogram with Dataset 45, 50, 52, 60, 61, 70, 72, 85, 90
y = [45, 50, 52, 60, 61, 70, 72, 85, 90] # Sample dataset
fig_test, ax_test = plt.subplots(figsize=(5,3)) # Create a new figure and axes
ax_test.hist(y) # Plots a histogram of the y values with 5 bins. The bins argument specifies the number of bins to use for the histogram.
```

```python
# Adding axis labels and titles
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50) 
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
#modify our labels and titles’ appearance by using font dictionaries (‘fontdict’)
fontdict1 = {'family': 'sans-serif', 'size': 18, 'color': 'red'}
fontdict2 = {'family': 'monospace', 'size': 12, 'color': 'black'}

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y)
ax.set_title('Total growth over time', fontdict=fontdict1)
ax.set_ylabel('Total growth', fontdict=fontdict2)
ax.set_xlabel('Years since start', fontdict=fontdict2)
fig.tight_layout()
```

```python
# Customize plot appearance
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y, 
           marker='*',
           color = "indigo")
fig.show()
```

```python
# We can also adjust the style and width of the line connecting our points

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y,
        marker = '*',
        color = 'indigo',
        linestyle = '--',
        linewidth = 1)
fig.show()
```

```python
# Modifying data points: Colour
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y,
           marker='*',
           color = "#3633FF", 
           linestyle = '--',
           linewidth = 1)
fig.show()
# Using a hex color code for a specific shade of orange 
```

```python
# Let’s customize our plot further
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y,
        marker = '*',
        color = "#7425b9",
        linestyle = '--',
        linewidth = 1,
        markeredgecolor = "#FA5981",
        markerfacecolor = '#000000',
        markersize = 10)
fig.show()
```

```python
# Create grid Lines and apply a style to them
fig, ax = plt.subplots(figsize=(5,3))
ax.grid(axis = 'y', color = 'red', linestyle = '--', linewidth = 1)
```
