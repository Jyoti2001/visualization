```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
sns.set_style("dark") # set the style of the plot to whitegrid
sns.lineplot(data=tips, # choose our dataset 
             x='total_bill', # define our x variable 
             y='tip') # define our y variable
```

```python
tipgraph = sns.lineplot(data=tips,
             x='total_bill',
             y='tip')
tipgraph.set_title("Total Bill vs Tip") # set the title of the graph
tipgraph.set_xlabel("Total Bill") # set the x-axis label
tipgraph.set_ylabel("Tip") # set the y-axis label
plt.show() # show the graph
```

```python
fig = plt.subplots(figsize=(10, 3))

tipgraph = sns.lineplot(data=tips, 
                        x='total_bill', 
                        y='tip', 
                        color = 'hotpink', 
                        linestyle = '--', 
                        linewidth = 3,
                        marker = 'o', 
                        markerfacecolor = 'indigo')
```

```python
#Seaborn makes it very easy to represent multiple variables with different visualelements of our graph

tipgraph = sns.scatterplot(data=tips, 
                           x='total_bill', 
                           y='tip', 
                           style = 'time',
                           hue = 'day', 
                           palette = ['purple', 'hotpink', 'deepskyblue', 'yellowgreen'])

tipgraph.set(title='Tips vs. Total Bill', xlabel='Total Bill ($)', ylabel='Tip Amount ($)')
```

```python
# pairplot is another convenientway to compare variables
sns.pairplot( data = tips, hue = 'day')
```

```python
# Replot
#relplot also lets us explorespecific levels within variables

daysplot = sns.relplot( data=tips, 
                       x="total_bill", 
                       y="tip", 
                       hue="sex", 
                       col="day", 
                       kind="scatter", 
                       col_wrap=2)
```

```python
#Plotly
import plotly.graph_objects as go
import nbformat
x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
# Plotly syntax is different from matplotlib/seaborn
graph = go.Figure()
```

```python
graph.add_trace(go.Bar(x=x1, y=y1))
graph.update_layout(title="Pirate Scores", 
                    xaxis_title="Pirates", 
                    yaxis_title="Scores")

graph.show()
```

```python
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, 
                           y=y1,
                           mode='markers',
                           marker = dict(color='hotpink', 
                                         opacity=1,
                                         size=12, 
                                         line=dict(color='indigo', width=5))))

graph.update_layout(title="Interactive Pirate Plot",
                    xaxis_title="Pirates", 
                    yaxis_title="Scores",
                    width=500,
                    height=500)
```

```python
#Wordclouds and venn diagrams
#Import the wordcloud package and load our sample dataset of movie quotes

from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",on_bad_lines='skip')
df
```

```python
text = " ".join(each for each in df.quote)

wordcloud = WordCloud(background_color='white',
                      colormap = 'inferno',
                      random_state = 613).generate(text)

fig, ax = plt.subplots(figsize=(10, 5))
ax.imshow(wordcloud)
ax.axis('off')
```

```python
#Making venn diagrams
# We can use the matplotlib_vennextension to make venn diagrams

from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A = set(['apple', 'banana', 'watermelon', 'date'])
B = set(['pumpkin', 'blueberry', 'apple', 'grape','key lime'])

diagram = venn2([A, B], set_labels = ('Fruits', 'Pies'),
                 set_colors = ('blue', 'red'), 
                 alpha = 0.5)

diagram.get_label_by_id('10').set_text("\n".join(A - B))

diagram.get_label_by_id('11').set_text("\n".join(A & B))

diagram.get_label_by_id('01').set_text("\n".join(B - A))
```
