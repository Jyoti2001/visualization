```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Create random data
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y)
fig.show()
```

```python
# Make our basic scatterplot:
fig, ax = plt.subplots(figsize=(5,3))

# Figure = set the entire canvas (in red) that contains everything
fig.set_facecolor('red') # changing the background color of the

# Axes = the plotting area inside a figure (the green boxes with x/y ticks where data isdrawn).
ax.set_facecolor('lightgreen') # changing the background color of the axes
ax.scatter(x, y)
fig.show()
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y1, label='Person 1')
ax.plot(x, y2, label='Person 2')
ax.legend()
```

```python
# Modifying our legens
# By adding arguments to our legen object, we can change different elements of our legend

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y1, label = "Person 1")
ax.plot(x, y2, label = "Person 2")
ax.legend(loc='lower right',
          frameon = True,
          fontsize = 12,
          ncol = 2,
          shadow = True)
fig.show()
```

```python
# Moving the legen outside of the the plot area
# We can Position the legend outside of the plot area by using the `bbox_to_anchor` argument. 
# This allows us to specify the exact location of the legend in relation to the plot area.

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y1, label = "Person 1")
ax.plot(x, y2, label = "Person 2")
ax.legend(loc='upper left', bbox_to_anchor=(1, 1))
fig.show()
```

```python
#Annotations with the text() function
# Lets start by putting our data into a scatter plot this time

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

# use the text() function to specify coordinates and text

ax.text(10,95, "This value is important!")
fig.show()
```

```python
#Modifying text annotations
# We can modify our text annotation by adding arguments

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y1, label ="Person1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.text(10, 95, "This Value is important !",
        ha = 'left',
        color = 'red',
        size = 10)
fig.show()
```

```python
# Positioning text annotations
# We can do this using transforms

fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, " * Data:(1, 5)", transform = ax.transData)
ax.text(0.5, 0.1, "* Axes:(0.5, 0.1)", transform = ax.transAxes)
ax.text(0.2, 0.2, "* Figure:(0.2, 0.2)", transform = fig.transFigure)
```

```python
# Annotating with arrows
# We can use the annotate() function to annotate with arrows and text (e.g to focus at a specific point)
fig, ax = plt.subplots()
ax.scatter(x, y1, label ="Person1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate("Important Point", xy=(5, 75), xytext=(3, 80),
            arrowprops=dict(arrowstyle='->', color='red'))
fig.show()
```

```python
# Removing tick marks and labels
# We can remove tick marks or labels from our axis like so:
# Use locator to remove ticks and labels
# Use formatter to remove only labels

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y1, label ="Person1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.NullLocator()) # remove ticks and labels
ax.yaxis.set_major_formatter(plt.NullFormatter()) # remove only labels
fig.show()
```

```python
# Limiting tick marks
# We can also define a maximum number of ticks we want on our axis

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y1, label ="Person1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(5)) 
```

```python
# Tick mark intervals
# We can set tick marks at a particular intervals (in this case 5) using MultipleLocator

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y1, label ="Person1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
fig.show()
```

```python
#Rotating axis labels
#To rotate axis labels and horizontal alignment, use xticks()
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y1, label ="Person1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45)
fig.show()
```

```python
# Modifying axis title fonts
font1 = {'family':'serif','color':'indigo'}
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xlabel('Shiny New X Axis!', fontsize = 18, fontdict = font1)
plt.ylabel('Shiny New Y Axis!', fontsize = 18, fontdict = font1)
fig.show()
```
